#guessing_game.py

```python
import numpy as np
number = np.random.randint(0,101)
guess = int(input("guess a number between 0 and 100: "))
while guess != number:
  if guess < number:
        print("The number to guess is higher.")
  else:
        print("The number to guess is lower.")
  guess = int(input("Try again: "))
print("You won!")
```
